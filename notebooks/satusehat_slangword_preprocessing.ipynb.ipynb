**LIBRARIES**

```python
import pandas as pd
from datasets import load_dataset
import string
```

**LOAD DATA**

```python
df = pd.read_csv("../data/data_raw_satusehat.csv")
ds = load_dataset("theonlydo/indonesia-slang")
```

**BUAT DICTIONARY**

```python
slang_df = ds["train"].to_pandas()
slang_dict = dict(zip(slang_df["slang"], slang_df["formal"]))
```

**SLANG TAMBAHAN**

```python
custom_slang = {
    "bbrapa": "beberapa",
    "dipke": "dipakai",
}
slang_dict.update(custom_slang)
```

Dictionary dari Hugging Face tidak lengkap. Beberapa slang words lokal seperti "bbrapa" dan "dipke" tidak ada di dataset tersebut. Oleh karena itu, kami tambahkan custom dictionary untuk slang words yang terlewat agar proses replace lebih sempurna.

**PREPROCESSING**

```python
def replace_slang(text):
    text_lower = text.lower()
    words = text_lower.split()
    result = []
    
    for word in words:
        clean_word = word.strip(string.punctuation)
        formal_word = slang_dict.get(clean_word, clean_word)
        result.append(formal_word)
    
    return " ".join(result)
```

```python
df["content"] = df["content"].apply(replace_slang)
```

**EXPORT DATASET**

```python
df.to_csv("../data/satusehat_reviews_slang_cleaned.csv", index=False)
```
